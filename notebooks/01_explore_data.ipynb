# 01 · Explore the dataset
Loads the processed splits, prints size/composition stats, and shows example (source, target) pairs.
Run `python scripts/prepare_data.py` first if `data/processed/*.jsonl` doesn't exist yet.

```
import sys, pathlib
sys.path.insert(0, str(pathlib.Path('..').resolve()))
from src.datasets.pipeline import load_split
from src.utils.io import read_json
import pandas as pd
PROCESSED = '../data/processed'
```

```
stats = read_json(f'{PROCESSED}/stats.json')
stats
```

## Split sizes

```
for name in ['train','valid','test','test_hard','test_unseen_words','gold_demo']:
    rows = load_split(PROCESSED, name)
    print(f'{name:20s} {len(rows):5d} rows')
```

## Sample pairs

```
train = load_split(PROCESSED, 'train')
df = pd.DataFrame(train)
df[['src','tgt','kind','source']].sample(10, random_state=1)
```

## Language-mix distribution (train)

```
import collections
mix_counts = collections.Counter(r['comp']['mix'] for r in train)
mix_counts
```
